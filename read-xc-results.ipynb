```python
from glob import glob
import numpy as np
```

```python
filelist = glob('*_xcorr.npz')
filelist.sort()
# filelist=filelist[0:2]

for i,f in enumerate(filelist):
    thisxc = np.load(f)['xc']
    #stack onto the arrays loaded from the previous files
    if i == 0:
        xcorr = thisxc
    else:
        xcorr = np.concatenate((xcorr, thisxc), axis=0)
    print(f, thisxc.shape, xcorr.shape)
```

```python
# max_xc = np.max(xcorr, axis=1)

import matplotlib.pyplot as plt
plt.rcParams['font.size'] = 20
plt.figure(figsize=(15, 5))


plt.imshow(xcorr, aspect='auto', origin = 'lower', cmap='viridis', vmin=0, vmax=0.1, interpolation = 'none') 
#
plt.colorbar(label='Correlation')
plt.ylabel('Correlation window number')
plt.xlabel('DAS Channel')
#plt.ylim(35,45)
plt.title('Correlation Across Channels')
plt.show()
```
